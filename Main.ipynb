```python
print("Top 10 countries")
fnd['country'].value_counts().head(10)
```

```python
gen = fnd['genres'].str.split(',')
gen = gen.explode()
gen = gen.str.strip()
print("Top 5 genres")
gen.value_counts().head(5)
```

```python
print("Top 5 rated movies")
top_movies = fnd[fnd['type'] == 'Movie'] \
    .sort_values(by='rating', ascending=False) \
    .head(5)
top_movies[['title', 'rating']]
```

```python
print("Top 5 rated series")
top_series = fnd[fnd['type'] == 'TV Show'] \
    .sort_values(by='rating', ascending=False) \
    .head(5)
top_series[['title', 'rating']]
```

```python
print("Top 5 rated Overall shows ")
print("NOTE : the top show may not match with either the top movie or series as the raking is just based on ratings and mutiple entries can have same rating and ties are not broken")
top_shows = fnd \
    .sort_values(by='rating', ascending=False) \
    .head(5)
top_shows[['title', 'rating']]
```

```python
print("We can get a better ranking based on both rating and voter count")
print("Top 5 Overall shows")
top_sh =fnd.sort_values(by=['rating','vote_count'],ascending = [False,False]).head(5)
top_sh[['title','rating','vote_count']]
```

```python
print('the show which made the most profit')
fnd['profits'] = fnd['revenue']-fnd['budget']
top_profit = fnd.sort_values(by='profits',ascending = False).head(1)
top_profit[['title','profits']]
```

```python
print("Top 10 countries with most 10 rated shows")
top = fnd[fnd['rating']==10]
top['country'].value_counts().head(10)
```

```python
print('Genres Vs mean rating')
fnd['genres'].str.split(',').explode().str.strip() \
    .to_frame().join(fnd['rating']) \
    .groupby('genres')['rating'].mean() \
    .sort_values(ascending=False).head(10)
```

```python
print("Top 3 shows with smallest description")
fnd['desc_length'] = fnd['description'].str.len()
ml=fnd.sort_values(by='desc_length', ascending=True).head(3)
ml[['title','desc_length']]
```

```python
!pip install matplotlib
import matplotlib.pyplot as plt
```

```python
country_counts = (
    top['country']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10)
)
country_counts.plot(kind='bar')

plt.title('Top Countries with 10-Rated Movies')
plt.xlabel('Country')
plt.ylabel('Number of Movies')
plt.xticks(rotation=90)
plt.show()
```

```python
fnd['budget_m'] = fnd['budget'] / 1000000
fnd['revenue_m'] = fnd['revenue'] / 1000000
plt.scatter(fnd['budget_m'], fnd['revenue_m'])
plt.title('Budget vs Revenue (in Millions)')
plt.xlabel('Budget (Million $)')
plt.ylabel('Revenue (Million $)')
plt.show()
```

```python
genr = (
    fnd['genres']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .head(10))
genr.plot(kind='pie')
plt.title('Top 10 Genres Distribution')
plt.ylabel('')
plt.show()
```
